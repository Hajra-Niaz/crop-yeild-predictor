```python
import pandas as pd
```

```python
df=pd.read_csv("crop_yield_prediction_1000.csv")
```

```python
# import encoder
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
```

```python
encoder
```

```python
df.columns
```

```python
df["Soil_Type"]=encoder.fit_transform(df["Soil_Type"])
```

```python
df.sample()
```

```python
# now split data
```

```python
x=df[['Rainfall_mm', 'Fertilizer_kg_per_hectare', 'Temperature_C','Soil_Type']]
y=df['Yield_ton_per_hectare']
```

```python
x.sample()
```

```python
y.sample()
```

```python
# now train data
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.sample()
```

```python
y_train.sample()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
import joblib as jb
```

```python
jb.dump(model,"crop.pkl")
```
